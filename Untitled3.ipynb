```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_olivetti_faces
from sklearn.preprocessing import StandardScaler
```

```python
# Load Olivetti Faces dataset
faces = fetch_olivetti_faces(shuffle=True, random_state=42)

# Select only 5 face images
X_images = faces.images[:5]

print("Number of images:", X_images.shape[0])
print("Image size:", X_images.shape[1], "x", X_images.shape[2])
```

```python
plt.figure(figsize=(12, 3))

for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(X_images[i], cmap='gray')
    plt.title(f"Face {i+1}")
    plt.axis('off')

plt.suptitle("Original Face Images")
plt.show()
```

```python
# Flatten each image
X = X_images.reshape(5, -1)

print("Original image shape:", X_images.shape)
print("Flattened data shape:", X.shape)
```

```python
# Calculate mean face
mean_face = np.mean(X, axis=0)

# Center the data
X_centered = X - mean_face

# Calculate covariance matrix
covariance_matrix = np.cov(X_centered, rowvar=False)

print("Mean face vector shape:", mean_face.shape)
print("Covariance matrix shape:", covariance_matrix.shape)
```

```python
# Calculate eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eigh(covariance_matrix)

# Sort eigenvalues from largest to smallest
indices = np.argsort(eigenvalues)[::-1]

eigenvalues = eigenvalues[indices]
eigenvectors = eigenvectors[:, indices]

# Select top 3
top_3_eigenvalues = eigenvalues[:3]
top_3_eigenvectors = eigenvectors[:, :3]

print("Top 3 Eigenvalues:")
print(top_3_eigenvalues)

print("\nTop 3 Eigenvectors shape:")
print(top_3_eigenvectors.shape)
```

```python
plt.figure(figsize=(10, 3))

for i in range(3):
    eigenface = top_3_eigenvectors[:, i].reshape(64, 64)

    plt.subplot(1, 3, i + 1)
    plt.imshow(eigenface, cmap='gray')
    plt.title(f"Eigenface {i+1}")
    plt.axis('off')

plt.suptitle("Top 3 Eigenfaces")
plt.show()
```

```python
# Project centered images onto top 3 eigenvectors
compressed_data = X_centered @ top_3_eigenvectors

print("Compressed data shape:", compressed_data.shape)
```

```python
# Reconstruct images using only top 3 eigenfaces
X_reconstructed = (
    compressed_data @ top_3_eigenvectors.T
) + mean_face

# Convert back to image format
reconstructed_images = X_reconstructed.reshape(5, 64, 64)
```

```python
plt.figure(figsize=(12, 6))

for i in range(5):
    # Original
    plt.subplot(2, 5, i + 1)
    plt.imshow(X_images[i], cmap='gray')
    plt.title(f"Original {i+1}")
    plt.axis('off')

    # Reconstructed
    plt.subplot(2, 5, i + 6)
    plt.imshow(reconstructed_images[i], cmap='gray')
    plt.title(f"Compressed {i+1}")
    plt.axis('off')

plt.suptitle("Image Compression using Top-3 Eigenfaces")
plt.tight_layout()
plt.show()
```

```python
original_values = 64 * 64
compressed_values = 3

print("Original values per image:", original_values)
print("Compressed values per image:", compressed_values)

compression_ratio = original_values / compressed_values

print("Conceptual compression ratio:",
      round(compression_ratio, 2), ": 1")
```
